```python
# LATIHAN 1: Operasi Dasar NumPy Array
import numpy as np

harga = np.array([5000, 7000, 3000, 12000, 4500])
print('Rata-rata harga:', harga.mean())
print('Harga tertinggi:', harga.max())
print('Harga setelah diskon 10%:', harga * 0.9)
```

**Analisis Latihan 1**

Perintah harga * 0.9 pada NumPy Array langsung mengalikan seluruh angka di dalamnya sekaligus. Pada List Python biasa, kodenya akan error karena list tidak bisa dikalikan dengan angka desimal.

```python
# LATIHAN 2: Membuat Series & DataFrame
import pandas as pd

data_kantin = {
    'menu': ['Nasi Goreng', 'Es Teh', 'Mie Ayam', 'Es Teh', None],
    'harga': [12000, 4000, 10000, 4000, 8000],
    'terjual': [23, 40, None, 35, 18]
}

df_latihan2 = pd.DataFrame(data_kantin)
print(df_latihan2)
```

**Analisis Latihan 2**

Data kosong berada di kolom menu (baris ke-4) dan terjual (baris ke-2). Jika tidak dibersihkan, hasil perhitungan statistik seperti rata-rata dan total pendapatan akan menjadi salah.


```python
# LATIHAN 3: Data Loading & Inspection
df = pd.read_csv('data_kantin - data_kantin.csv')

print(df.head())
print(df.info())
print(df.describe())
print(df.shape)
```

**Analisis Latihan 3**

Kolom menu dan terjual punya jumlah data terisi yang lebih sedikit dari total 110 baris. Ini menandakan ada data kosong (missing value) pada kedua kolom tersebut.

```python
# LATIHAN 4: Menangani Missing Value

print(df.isnull().sum())
df['terjual'] = df['terjual'].fillna(0)
df = df.dropna(subset=['menu'])
```

**Analisis Latihan 4**

Kolom terjual diisi angka 0 (fillna) karena diasumsikan tidak ada transaksi penjualan. Baris dengan menu kosong dihapus (dropna) agar tidak menebak nama produk yang membuat laporan tidak akurat.

```python
# LATIHAN 5: Menangani Duplikat dan Tipe Data

print(df.duplicated().sum())
df = df.drop_duplicates()

df['harga'] = df['harga'].astype(int)
print(df.dtypes)
```

**Analisis Latihan 5**

Jumlah data berkurang dari 110 menjadi 102 baris setelah duplikat dihapus. Memastikan tipe data benar sangat penting agar perhitungan angka keuangan berjalan tepat tanpa error.

```python
# LATIHAN 6: Data Manipulation (Filtering, Sorting, Groupby)
laris = df[df['terjual'] > 20]
urut = df.sort_values(by='terjual', ascending=False)

df['total_pendapatan'] = df['harga'] * df['terjual']

ringkasan = df.groupby('menu')['total_pendapatan'].sum() # agregasi
print(ringkasan)
```

**Analisis Latihan 6**

Menu dengan pendapatan tertinggi adalah Nasi Goreng dengan total omzet Rp 3.924.000. Informasi ini berguna untuk menjaga stok bahan baku Nasi Goreng dan membuat strategi paket promo bundling.
